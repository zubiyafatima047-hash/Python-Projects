```python
import qrcode 
url=input("Enter URL to Make QR: ")
if url.startswith("http://") or url.startswith("https://"):
    qr=qrcode.QRCode(version=1,
                     error_correction=qrcode.constants.ERROR_CORRECT_H,
                     box_size=10,
                     border=4
                    )
    qr.add_data(url)
    qr.make()

    image=qr.make_image(fill_colors="blue", back_colors="red")

    image.save("ThirdQr.png")

else:
    print("Invalid Url..!!! Must start with http or https")
```

```python
from PIL import Image, ImageDraw, ImageFont
txt=input("Enter Your Text: ")

#create Image
image=Image.new("RGB", (800,300), "white")

#create drawing object
draw=ImageDraw.Draw(image)

#load Font
font=ImageFont.load_default()

#set left and right margins
left_margin=50
right_margin=50

#calculate width
available_width=800-left_margin-right_margin

#break text into mutiline
words=txt.split()
wrapped_lines=[]
current_line=""

for word in words:
    test_line=current_line + " "+word if current_line else word
    text_width= draw.textbbox((0,0), test_line, font=font)[2]

    if text_width<=available_width:
        current_line=test_line
    else:
        if current_line:
            wrapped_lines.append(current_line)
        current_line=word

if current_line:
    wrapped_lines.append(current_line)

#Draw text line by line 
y=50

for line in wrapped_lines:
    draw.text((left_margin,y), line, fill="red", font=font)
    y+=25

image.save("test_sample_24.png")
print("Image Generated Succesfully!")
```
